### PERGUNTA 1: Qual é a composição do nosso dataset? (Distribuição do Alvo)
#### ==============================================================================

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(20, 8))
os.makedirs(os.path.join(FIGURES_PATH, 'pergunta01'), exist_ok=True)

# Gráfico 1: Distribuição Binária (Geral)
sns.countplot(
    data=df, 
    x='diagnosed_diabetes', 
    palette='pastel',
    ax=ax1
)
ax1.set_title('Distribuição Binária (Geral)', fontsize=16)
ax1.set_xlabel('Diabetes Diagnosticado', fontsize=12)
ax1.set_ylabel('Contagem de Pacientes', fontsize=12)
ax1.set_xticklabels(['Não (0)', 'Sim (1)'])

for p in ax1.patches:
    ax1.annotate(f'{p.get_height():,}\n({p.get_height()/len(df):.1%})', 
                (p.get_x() + p.get_width() / 2., p.get_height()), 
                ha='center', va='center', xytext=(0, 10), textcoords='offset points')

# Gráfico 2: Distribuição Detalhada (Estágios)
sns.countplot(
    data=df, 
    x='diabetes_stage', 
    order=df['diabetes_stage'].value_counts().index,
    palette='Set1',
    ax=ax2
)

for p in ax2.patches:
    ax2.annotate(f'{p.get_height():,}\n({p.get_height()/len(df):.1%})', 
                (p.get_x() + p.get_width() / 2., p.get_height()), 
                ha='center', va='center', xytext=(0, 10), textcoords='offset points')

ax2.set_title('Distribuição Detalhada (Estágios)', fontsize=16)
ax2.set_xlabel('Estágio do Diabetes', fontsize=12)
ax2.set_ylabel('Contagem de Pacientes', fontsize=12)
ax2.tick_params(axis='x', rotation=45)


plt.suptitle('Distribuição da Classificação de Diabetes', fontsize=20, y=1.03)
sns.despine(fig=fig)
plt.tight_layout()

plt.savefig(os.path.join(FIGURES_PATH, 'pergunta01', 'distribuicao_alvo_combinada.png'), dpi=300, bbox_inches='tight')
plt.show()
```

```python
print("Tabela de Correlação")

tabela_consistencia = pd.crosstab(
    df['diagnosed_diabetes'],
    df['diabetes_stage'] 
)

display(tabela_consistencia)
```

Propósito: realizar um diagnóstico da qualidade dos dados da variável-alvo (diagnosed_diabetes), garantindo que a variável utilizada nas etapas subsequentes de modelagem reflita o estado clínico correto e não introduza erros sistemáticos no treinamento do modelo.

Insight: descoberta de uma inconsistência crítica nos dados. Embora a distribuição binária mostre 60% de casos positivos e a distribuição detalhada mostre 31.8$ de Pré-Diabetes, a tabela cruzada revela que 31.845 casos de Pré-Diabetes e 56 casos de Diabetes Tipo 1 estão incorretamente rotulados como 'Não Diabetes (0)' na variável binária. Esta falha de rotulagem compromete a integridade da variável-alvo original.

A partir disso, nota-se que alguns casos da diabete gestacional não é diagnosticada como diabetes, talvez seja por que na maioria dos casos é temporária;

Ação/Decisão: quando for trabalhado modelos preditivos coluna diagnosed_diabetes deve ser descartada devido à sua inconsistência. A ação central é criar uma nova variável-alvo binária a partir de diabetes_stage, classificando 'Pré-Diabetes', 'Type 1', 'Type 2' (e possivelmente 'Gestational', dependendo do foco do modelo em risco crônico) como 'Sim (1)', e apenas 'No Diabetes' como 'Não (0)', garantindo a precisão clínica.

### PERGUNTA 2: Qual a relação entre os exames de sangue e o risco de diabetes?
#### ==============================================================================

```python
df_subset = df[df['diabetes_stage'].isin(['No Diabetes', 'Pre-Diabetes', 'Type 2'])]
os.makedirs(os.path.join(FIGURES_PATH, 'pergunta02'), exist_ok=True)

sns.scatterplot(
    data=df_subset,
    x='hba1c',
    y='glucose_fasting',
    hue='diabetes_stage',
    palette='bright',
    alpha=0.6
)
plt.title('Glicose em Jejum vs. Hemoglobina Glicada (HbA1c)', fontsize=16)
plt.xlabel('HbA1c (%)', fontsize=12)
plt.ylabel('Glicose em Jejum (mg/dL)', fontsize=12)
plt.legend(title='Estágio')
plt.savefig(os.path.join(FIGURES_PATH, 'pergunta02', 'scatter_hba1c_vs_glicose.png'), dpi=300, bbox_inches='tight')
plt.show()
```

Propósito: validar a eficácia e a coerência dos exames de Glicose em Jejum e HbA1c como ferramentas de diagnóstico e rastreamento, confirmando as fronteiras clínicas utilizadas para classificar o risco do paciente

Insight: A relação entre Glicose em Jejum e HbA1c é linearmente positiva e valida as zonas de risco clínico: Não Diabetes (baixo risco) abaixo de 100 mg/dL e 5.7%, Pré-Diabetes (risco intermediário/transição) entre 100-125 mg/dL e 5.7-6.4%, e Tipo 2 (alto risco) acima desses limiares. A sobreposição nas fronteiras de classificação indica que o diagnóstico exige a análise conjunta de ambos os exames.

Ação/Decisão: Manter o rastreamento duplo (Glicose em Jejum e HbA1c) como padrão por ser mais confiável. Priorizar o treinamento de equipes para focar em pacientes nas zonas de fronteira (100 mg/dL ou  5.7%) para intervenção preventiva imediata. Em modelos preditivos, Glicose em Jejum e HbA1c são features críticas e devem ter o maior peso.

### PERGUNTA 3: Qual é a distribuição exata de idade para pacientes com diabetes?
#### ==============================================================================

```python
plt.figure(figsize=(12, 7))
os.makedirs(os.path.join(FIGURES_PATH, 'pergunta03'), exist_ok=True)

palette = {
    'No Diabetes': '#008000',
    'Pre-Diabetes': '#FF8C00',
    'Type 2': '#B22222',
    'Type 1': '#1E90FF',
    'Gestational': '#8A2BE2'
}

for stage, color in palette.items():
    subset = df[df['diabetes_stage'] == stage]
    sns.kdeplot(
        data=subset,
        x='age',
        fill=True,
        alpha=0.35,
        linewidth=2,
        color=color,
        label=stage
    )

plt.title('Distribuição da Idade por Estágio do Diabetes', fontsize=16, weight='bold')
plt.xlabel('Idade do Paciente', fontsize=12)
plt.ylabel('Densidade Estimada', fontsize=12)
plt.legend(title='Estágio do Diabetes', fontsize=10, title_fontsize=11)
plt.tight_layout()
plt.savefig(os.path.join(FIGURES_PATH, 'pergunta03', 'kde_idade_por_estagio_diabetes.png'), dpi=300, bbox_inches='tight')
plt.show()
```

Propósito:mapear a idade de maior risco para cada tipo de diabetes

Insight:A idade de diagnóstico varia acentuadamente: o Diabetes Tipo 1 concentra-se em jovens adultos (25-30 anos) e o Gestacional em idade fértil (35-40 anos). Já o Tipo 2 e o Pré-Diabetes dominam as distribuições em adultos mais velhos (60-65 anos), confirmando que o risco se desloca para doenças crônicas com o avanço da idade.

Ação/Decisão
O rastreamento deve ser focado no Tipo 1 para jovens (menores de 40) e na gestão crônica de Pré-Diabetes/Tipo 2 para a meia-idade e idosos (45+). O cuidado materno deve otimizar a triagem gestacional na idade fértil (30-45). Em modelos de aprendizagem de máquina, a Idade deve ser tratada de forma não linear.

### PERGUNTA 06: Qual o perfil do diabetes risk score médio para cada faixa etária da população?
#### ==============================================================================

```python
os.makedirs(os.path.join(FIGURES_PATH, 'pergunta06'), exist_ok=True)

df['age_group'] = pd.cut(
    df['age'],
    bins=[17,30,40,50,60,70,90],
    labels=['18-30','31-40','41-50','51-60','61-70','71+']
)

risk_mean_age = df.groupby('age_group')['diabetes_risk_score'].mean().reset_index()

heat_data = (
    df.groupby('age_group')['diabetes_stage']
      .value_counts(normalize=True)
      .rename('proportion')
      .reset_index()
      .pivot(index='age_group', columns='diabetes_stage', values='proportion')
)


fig, axes = plt.subplots(1, 2, figsize=(15, 5))

# Gráfico 1: lineplot
sns.lineplot(
    data=risk_mean_age,
    x='age_group',
    y='diabetes_risk_score',
    marker='o',
    linewidth=2,
    color='#ff8c00',
    ax=axes[0]
)

axes[0].set_title('Média do Risco de Diabetes por Faixa Etária', fontsize=13, weight='bold')
axes[0].set_xlabel('Faixa Etária')
axes[0].set_ylabel('Risco Médio (0–100)')
axes[0].grid(alpha=0.3)

# Gráfico 2: heatmap
sns.heatmap(
    heat_data,
    annot=True,
    cmap='YlOrRd',
    fmt='.2f',
    linewidths=.5,
    cbar_kws={'label': 'Proporção'},
    ax=axes[1]
)

axes[1].set_title('Proporção de Estágios por Faixa Etária', fontsize=13, weight='bold')
axes[1].set_xlabel('Estágio do Diabetes')
axes[1].set_ylabel('Faixa Etária')

plt.tight_layout()
plt.savefig(os.path.join(FIGURES_PATH, 'pergunta06', 'lineplot_e_heatmap_faixa_etaria.png'),dpi=300, bbox_inches='tight')
plt.show()
```

Propósito: quantificar a progressão do risco de diabetes ao longo do ciclo de vida e identificar as faixas etárias de maior necessidade de intervenção preventiva versus tratamento ativo.

Insight: O Risco Médio aumenta consistentemente com cada faixa etária, e essa escalada é impulsionada pela transição progressiva dos pacientes de estágios de "Pré-Diabetes" para o estágio de "Tipo 2".

Ação/Decisão: Focar a prevenção (estilo de vida) em faixas jovens/intermediárias (31-60), onde o Pré-Diabetes é alto, e a gestão ativa/tratamento em idosos (61+), dominados pelo Tipo 2. A Idade é o fator preditivo mais crítico para modelagem em algoritmos de aprendizagem de máquina.

### PERGUNTA 07: Ter histórico familiar ou hipertensão aumenta o risco médio de diabetes?
#### ==============================================================================

```python
from sklearn.preprocessing import MinMaxScaler

os.makedirs(os.path.join(FIGURES_PATH, 'pergunta07'), exist_ok=True)

# Criando grupos clínicos combinados
df['historico_grupo'] = df.apply(
    lambda x: 
        'Nenhum' if (x['family_history_diabetes'] == 0 and x['hypertension_history'] == 0)
        else 'Apenas Familiar' if (x['family_history_diabetes'] == 1 and x['hypertension_history'] == 0)
        else 'Apenas Hipertensão' if (x['family_history_diabetes'] == 0 and x['hypertension_history'] == 1)
        else 'Ambos'
    , axis=1
)

# Cálculo do risco médio
risk_hist = df.groupby('historico_grupo')['diabetes_risk_score'].mean().reset_index()

# Ordenando do menor para maior risco
risk_hist = risk_hist.sort_values(by='diabetes_risk_score')

plt.figure(figsize=(8,5))
sns.barplot(
    data=risk_hist,
    x='diabetes_risk_score',
    y='historico_grupo',
    palette='Reds'
)

plt.title('Risco Médio de Diabetes por Histórico Clínico', fontsize=14, weight='bold')
plt.xlabel('Risco Médio (0 a 100)')
plt.ylabel('Grupo Clínico')
plt.tight_layout()
plt.savefig(os.path.join(FIGURES_PATH, 'pergunta07', 'risco_historico_clinico.png'), dpi=300, bbox_inches='tight')
plt.show()
```

Propósito: quantificar e comparar o Risco Médio de Diabetes entre pacientes com diferentes combinações de históricos clínicos (familiar e hipertensão), permitindo identificar os grupos de maior e menor vulnerabilidade.

Insight: A principal descoberta é que o Histórico Familiar tem um impacto muito maior no Risco de Diabetes do que o Histórico de Hipertensão, e ter Ambos os fatores de risco não aumenta o risco de forma substancial em relação a ter Apenas Familiar.

Ações/Decisões: Priorizar rastreamento e intervenções intensivas (programas de estilo de vida) em pacientes com Histórico Familiar de Diabetes e atribuir maior peso a essa feature em modelos preditivos, visto que ela é o principal fator de risco, minimizando o foco exclusivo na Hipertensão.

### PERGUNTA 08: Qual combinação entre gênero e faixa etária apresenta maior risco médio de diabetes?
#### ==============================================================================

```python
os.makedirs(os.path.join(FIGURES_PATH, 'pergunta08'), exist_ok=True)

# Cálculo da média por faixa etária e gênero
heat = df.groupby(['age_group', 'gender'])['diabetes_risk_score'].mean().reset_index()

# Pivot para heatmap
heat_pivot = heat.pivot(index='age_group', columns='gender', values='diabetes_risk_score')

plt.figure(figsize=(7,5))
sns.heatmap(
    heat_pivot,
    annot=True,
    fmt='.1f',
    cmap='coolwarm',
    linewidths=0.5,
    linecolor='gray'
)

plt.title('Mapa de Calor do Risco de Diabetes por Idade e Gênero', fontsize=14, weight='bold')
plt.xlabel('Gênero')
plt.ylabel('Faixa Etária')
plt.tight_layout()
plt.savefig(os.path.join(FIGURES_PATH, 'pergunta08', 'heatmap_idade_genero_risco.png'), dpi=300, bbox_inches='tight')
plt.show()
```

Propósito: quantificar e isolar o impacto da idade e do gênero no risco de diabetes

Insight: o fator etário é o preditor dominante e mais influente no risco de diabetes, com a incidência aumentando drasticamente e progressivamente com a idade, passando de aproximadamente $22-23\%$ na faixa $18-30$ para cerca de $38\%$ na faixa $71+$. O Gênero (Female, Male, Other) apresenta uma diferença de risco insignificante ou quase nula dentro de qualquer faixa etária específica.

Ação/Decisão: Concentrar programas de rastreamento, prevenção e alocação de recursos de saúde nas faixas etárias mais elevadas (51+), pois é onde o risco é mais concentrado, independentemente do gênero do paciente. Em modelos de aprendizagem de máquina, atribuir alto peso à Idade e baixo peso ao Gênero.
