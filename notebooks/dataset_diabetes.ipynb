```python
categorical_cols = ['gender', 'ethnicity', 'education_level', 'income_level',
                        'employment_status', 'smoking_status', 'diabetes_stage']
for col in categorical_cols:
    if col in df.columns:
        print(f"\nValores em '{col}':")
        print(df[col].value_counts(dropna=False))
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.countplot(x='gender', hue='diagnosed_diabetes', data=df)
plt.title("Diabetes por Gênero")
plt.show()
```

```python

numerical_features = [
    'age', 'alcohol_consumption_per_week', 'physical_activity_minutes_per_week', 
    'diet_score', 'sleep_hours_per_day', 'screen_time_hours_per_day', 
    'bmi', 'waist_to_hip_ratio', 'systolic_bp', 'diastolic_bp', 
    'heart_rate', 'cholesterol_total', 'hdl_cholesterol', 
    'ldl_cholesterol', 'triglycerides', 'glucose_fasting', 
    'glucose_postprandial', 'insulin_level', 'hba1c', 'diabetes_risk_score'
]

plt.figure(figsize=(20, 25)) 

for i, col in enumerate(numerical_features, 1):
    if col in df.columns:
        plt.subplot(5, 4, i) 
        sns.boxplot(y=df[col])
        plt.title(f'Boxplot de {col}')
    
plt.tight_layout()
plt.show()

# plt.savefig('todos_os_boxplots.png')
```

```python
if not df.empty:
   
    numeric_df_for_corr = df.select_dtypes(include=['float64', 'int64'])
    
    plt.figure(figsize=(20, 15))
    correlation_matrix = numeric_df_for_corr.corr()
    sns.heatmap(correlation_matrix, annot=True, fmt='.2f', cmap='coolwarm', linewidths=0.5)
    plt.title('Matriz de Correlação de Pearson')
    plt.show()

    if 'diagnosed_diabetes' in correlation_matrix:
        print("\n--- Correlação com o Diagnóstico de Diabetes ---")
        print(correlation_matrix['diagnosed_diabetes'].sort_values(ascending=False))
```

```python
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
sns.boxplot(data=df, x='diagnosed_diabetes', y='hba1c')
plt.title('Hba1c (Hemoglobina Glicada) vs. Diagnóstico')

plt.subplot(1, 2, 2)
sns.boxplot(data=df, x='diagnosed_diabetes', y='glucose_fasting')
plt.title('Glicose em Jejum vs. Diagnóstico')
plt.suptitle('Pergunta 1: Impacto dos Principais Biomarcadores no Diagnóstico', y=1.03)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 6))
sns.scatterplot(data=df, x='glucose_fasting', y='hba1c', hue='diagnosed_diabetes', alpha=0.6)
plt.title('Glicose em Jejum vs. Hba1c (Curto vs. Longo Prazo)')
plt.xlabel('Glicose em Jejum (mg/dL)')
plt.ylabel('Hba1c (%)')
plt.show()
```

```python
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
sns.scatterplot(data=df, x='physical_activity_minutes_per_week', y='bmi', alpha=0.5)
plt.title('Atividade Física vs. IMC')

plt.subplot(1, 2, 2)
sns.boxplot(data=df, x='diet_score', y='bmi')
plt.title('Pontuação da Dieta (0-10) vs. IMC')
plt.suptitle('Pergunta 3: Estilo de Vida e seu Impacto no IMC', y=1.03)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
sns.countplot(data=df, x='hypertension_history', hue='diagnosed_diabetes', palette='viridis')
plt.title('Diagnóstico de Diabetes por Histórico de Hipertensão')
plt.xlabel('Hipertensão (0=Não, 1=Sim)')

plt.subplot(1, 2, 2)
sns.countplot(data=df, x='cardiovascular_history', hue='diagnosed_diabetes', palette='plasma')
plt.title('Diagnóstico de Diabetes por Histórico Cardiovascular')
plt.xlabel('Cardiovascular (0=Não, 1=Sim)')
plt.suptitle('Pergunta 4: Risco de Comorbidades (Hipertensão e Cardio)', y=1.03)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(7, 5))
sns.countplot(data=df, x='family_history_diabetes', hue='diagnosed_diabetes')
plt.title('Impacto do Histórico Familiar de Diabetes no Diagnóstico')
plt.xlabel('Histórico Familiar (0=Não, 1=Sim)')
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.violinplot(data=df, x='diabetes_stage', y='age', order=['No Diabetes', 'Pre-Diabetes', 'Type 1', 'Type 2', 'Gestational'])
plt.title('Pergunta 6: Distribuição de Idade por Estágio do Diabetes')
plt.xlabel('Estágio do Diabetes')
plt.ylabel('Idade')
plt.xticks(rotation=15)
plt.show()
```

```python
income_diabetes_ct = pd.crosstab(df['income_level'], df['diagnosed_diabetes'])
# Normalizar para ver porcentagens
income_diabetes_pct = income_diabetes_ct.div(income_diabetes_ct.sum(axis=1), axis=0)

plt.figure(figsize=(14, 6))
plt.subplot(1, 2, 1)
income_diabetes_pct.plot(kind='bar', stacked=True, ax=plt.gca())
plt.title('Proporção de Diabetes por Nível de Renda')
plt.xlabel('Nível de Renda')
plt.ylabel('Proporção')
plt.xticks(rotation=0)

plt.subplot(1, 2, 2)
sns.boxplot(data=df, x='income_level', y='diet_score', order=['Low', 'Medium', 'High'])
plt.title('Pontuação da Dieta por Nível de Renda')
plt.xlabel('Nível de Renda')
plt.ylabel('Pontuação da Dieta')

plt.suptitle('Pergunta 7: Fatores Socioeconômicos e Saúde', y=1.03)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(14, 6))
plt.subplot(1, 2, 1)
sns.boxplot(data=df, x='smoking_status', y='bmi')
plt.title('IMC por Hábito de Fumar')

plt.subplot(1, 2, 2)
# Agrupar tempo de tela para visualização (ou usar scatter)
df['screen_time_bins'] = pd.cut(df['screen_time_hours_per_day'], bins=[0, 2, 4, 6, 8, 12], right=False)
sns.boxplot(data=df, x='screen_time_bins', y='bmi')
plt.title('IMC por Tempo de Tela Diário')
plt.xlabel('Horas de Tempo de Tela')

plt.suptitle('Pergunta 8: Comportamentos e IMC', y=1.03)
plt.tight_layout()
plt.show()
```
